Create 1D NumPy array

```python
import numpy as np

arr = np.arange(1, 21)
print(arr)
```

Calculate the sum, mean, median, and standard deviation using build-in function

```python
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Std Dev:", np.std(arr))
```

Find the indices of elements greater than 10

```python
indices = np.where(arr > 10)
print("Indices:", indices)
```

create 2D NumPy array of shape 4 X 4

```python
arr2 = np.arange(1, 17).reshape(4, 4)
print(arr2)
```

taking out transpose

```python
print("Transpose:\n", arr2.T)
```

row-wise and column-wise sums

```python
print("Row-wise sum:", np.sum(arr2, axis=1))
print("Column-wise sum:", np.sum(arr2, axis=0))
```

Create two 3 X 3 arrays

```python
a = np.random.randint(1, 21, (3, 3))
b = np.random.randint(1, 21, (3, 3))

print("A:\n", a)
print("B:\n", b)
```

addition, subtraction, and multiplication of two 2D array

```python
print("Addition:\n", a + b)
print("Subtraction:\n", a - b)
print("Multiplication:\n", a * b)
```

dot product of the two arrays

```python
print("Dot Product:\n", np.dot(a, b))
```

Reshape a 1D array of size 12 into a 3 X 4 2D array and slice the first two rows and last two columns.

```python
arr3 = np.arange(1, 13).reshape(3, 4)
print(arr3)
```

slice the first two rows and last two columns

```python
slice_part = arr3[:2, 2:]
print(slice_part)
```
