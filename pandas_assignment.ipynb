create a dataFrame

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

df = pd.DataFrame(data)
print(df.head())
```

Get the summary statistics of the 'Age' and 'Salary' columns

```python
print(df[['Age', 'Salary']].describe())
```

Calculate the average salary of employees

```python
avg_hr_salary = df[df['Department'] == 'HR']['Salary'].mean()
print(avg_hr_salary)
```

Add a new column of 'Bonus', which is 10% of the salary

```python
df['Bonus'] = df['Salary'] * 0.10
print(df)
```

Filter the DataFrame to show employees age between 25 and 30

```python
filtered_df = df[(df['Age'] >= 25) & (df['Age'] <= 30)]
print(filtered_df)
```

Group the data by 'Department' and calculate the average salary for each department

```python
grouped = df.groupby('Department')['Salary'].mean()
print(grouped)
```

Sort the DataFrame by 'Salary' in ascending order and save the result to a new CSV file

```python
sorted_df = df.sort_values(by='Salary', ascending=True)
sorted_df.to_csv('sorted_employees.csv', index=False)
```
