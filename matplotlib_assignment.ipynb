Visualize how y changes with respect to x.
We plot the values using plot()

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,25,30,50]

plt.figure()
plt.plot(x, y)
plt.title('Line Plot of Y vs X')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

Goal

Compare marks scored by students.

Explanation

We use bar() to display categories (students) vs values (marks)

Added:
✔ Title
✔ Labels

```python
import matplotlib.pyplot as plt

students = ['John','Jane','Alice','Bob']
marks = [75,85,60,90]

plt.figure()
plt.bar(students, marks)
plt.title('Student Marks')
plt.xlabel('Students',fontdict={'color': 'yellow'})
plt.ylabel('Marks', fontdict={'color': 'brown'})
plt.show()
```

Show percentage revenue from regions and highlight highest region.

We use pie()/
Explode effect is used to highlight the largest slice/
autopct displays percentages

```python
import matplotlib.pyplot as plt

regions = ['North America','Europe','Asia','Others']
revenue = [45,25,20,10]

explode = [0.1 if r==max(revenue) else 0 for r in revenue]

plt.figure()
plt.pie(revenue, labels=regions, explode=explode, autopct='%1.1f%%')
plt.title('Revenue Distribution by Region')
plt.show()
```

Show frequency distribution of 1000 random integers (1–100).

We generate data using random.randint()

Use hist() to show how often values appear

Histogram helps understand distribution patterns

```python
import matplotlib.pyplot as plt
import random

data = [random.randint(1,100) for i in range(1000)]

plt.figure()
plt.hist(data, bins=20)
plt.title('Frequency Distribution of Random Integers')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```
